# Figure 8 of Pierard2024TheTile: Grid

```python
from sorbetto.tile import Tile
from sorbetto.parameterization import ParameterizationDefault
from sorbetto.annotation import AnnotationText, AnnotationGridShiftClassPriors, AnnotationCurveFixedClassPriors
from sorbetto.ranking import RankingScore

import matplotlib.pyplot as plt
```

```python
def produce_subfigure ( prior_pos ) :

    prior_pos_old = 0.5
    prior_pos_new = prior_pos
    
    parameterization = ParameterizationDefault ()
    tile = Tile (parameterization, name='Tile for $\\pi_+={:g}$'.format ( prior_pos_new ) )
    
    annotation = AnnotationGridShiftClassPriors (prior_pos_old, prior_pos_new, linestyle='-' )
    tile.appendAnnotation ( annotation )
    
    annotation = AnnotationCurveFixedClassPriors (prior_pos_new, color='tab:blue' )
    tile.appendAnnotation ( annotation )
    
    score = RankingScore.getAccuracy ()
    annotation = AnnotationText ( score, color='tab:green', fontsize=12 )
    tile.appendAnnotation ( annotation )
    
    score = RankingScore.getBalancedAccuracy ( prior_pos_new )
    annotation = AnnotationText ( score, color='tab:purple', fontsize=12 )
    tile.appendAnnotation ( annotation )
    
    tile.draw ()
    plt.show ()
    
```

## Shift from $(\pi_−,\pi_+) = (0.5,0.5)$ to $(0.9,0.1)$

```python
produce_subfigure ( 0.1 )
```

## Shift from $(\pi_−,\pi_+) = (0.5,0.5)$ to $(0.7,0.3)$

```python
produce_subfigure ( 0.3 )
```
