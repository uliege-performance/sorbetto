# Figure 6 of Pierard2024TheTile: The "No-Skills Equivalence Flavor"

```python
from sorbetto.flavor import NoSkillsEquivalenceFlavor
from sorbetto.parameterization import ParameterizationDefault
from sorbetto.ranking import RankingScore
from sorbetto.tile import NumericTile
from sorbetto.annotation import AnnotationIsovalueCurves, AnnotationCurveFixedClassPriors, AnnotationCurveFixedPredictionRates, AnnotationText

import matplotlib.pyplot as plt
```

## Displaying the constraint on $\tau_-$ on a Tile as well as some $\gamma_\tau$ curves on top of it

```python
flavor = NoSkillsEquivalenceFlavor ( 'negative prediction rate' )
parameterization = ParameterizationDefault ()
tile = NumericTile ( flavor=flavor, parameterization=parameterization )
tile.appendAnnotation ( AnnotationIsovalueCurves () )
ratePos = 0.3
tile.appendAnnotation ( AnnotationCurveFixedPredictionRates ( ratePos=ratePos, color='tab:orange' ) )
tile.appendAnnotation ( AnnotationText ( RankingScore.getMacroAveragedPrecision ( ratePos=ratePos ), color='tab:orange' ) )
tile.draw ()
plt.show ()
```

## Displaying the constraint on $\tau_+$ on a Tile as well as some $\gamma_\tau$ curves on top of it

```python
flavor = NoSkillsEquivalenceFlavor ( 'positive prediction rate' )
parameterization = ParameterizationDefault ()
tile = NumericTile ( flavor=flavor, parameterization=parameterization )
tile.appendAnnotation ( AnnotationIsovalueCurves () )
ratePos = 0.3
tile.appendAnnotation ( AnnotationCurveFixedPredictionRates ( ratePos=ratePos, color='tab:orange' ) )
tile.appendAnnotation ( AnnotationText ( RankingScore.getMacroAveragedPrecision ( ratePos=ratePos ), color='tab:orange' ) )
tile.draw ()
plt.show ()
```

## Displaying the constraint on $\pi_-$ on a Tile as well as some $\gamma_\pi$ curves on top of it

```python
flavor = NoSkillsEquivalenceFlavor ( 'prior of the negative class' )
parameterization = ParameterizationDefault ()
tile = NumericTile ( flavor=flavor, parameterization=parameterization )
tile.appendAnnotation ( AnnotationIsovalueCurves () )
priorPos = 0.3
tile.appendAnnotation ( AnnotationCurveFixedClassPriors ( priorPos=priorPos, color='tab:orange' ) )
tile.appendAnnotation ( AnnotationText ( RankingScore.getMacroAveragedRecall ( priorPos=priorPos ), color='tab:orange' ) )
tile.draw ()
plt.show ()
```
