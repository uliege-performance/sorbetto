```python
from sorbetto.flavor import NoSkillsEquivalenceFlavor
from sorbetto.core import Importance
from sorbetto.parameterization import ParameterizationDefault
from sorbetto.ranking import RankingScore
from sorbetto.tile import NumericTile
from sorbetto.annotation import AnnotationIsovalueCurves, AnnotationCurveFixedClassPriors, AnnotationCurveFixedPredictionRates
import random
import matplotlib.pyplot as plt
import numpy as np
```

```python
itn = random.random ()
ifp = random.random ()
ifn = random.random ()
itp = random.random ()
importance = Importance ( itn, ifp, ifn, itp )

parameterization = ParameterizationDefault ()
score = RankingScore ( importance )
a = parameterization.getValueParameter1 ( score )
b = parameterization.getValueParameter2 ( score )
```

```python
flavor = NoSkillsEquivalenceFlavor ( 'negative prediction rate' )
print ( flavor ( importance ) )
#DEPRECATED: print ( ParameterizationDefault.getRateNegPredictionsForIsoValuedNoSkillPerformances ( a, b ) )
tile = NumericTile ( flavor=flavor, parameterization=parameterization )
tile.appendAnnotation ( AnnotationIsovalueCurves () )
tile.appendAnnotation ( AnnotationCurveFixedPredictionRates ( ratePos=0.3, color='tab:orange' ) )
tile.draw ()
plt.show ()
print ( 'min:', tile.minimize () )
print ( 'max:', tile.maximize () )
```
