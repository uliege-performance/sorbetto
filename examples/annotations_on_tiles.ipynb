# Annotation on Tiles

```python
from sorbetto.tile import Tile
from sorbetto.parameterization import ParameterizationDefault
from sorbetto.annotation import AnnotationText, AnnotationImportanceCompass
from sorbetto.ranking import RankingScore

import matplotlib.pyplot as plt
```

## Drawing the "Importance Compass" Annotation

```python
parameterization = ParameterizationDefault ()
tile = Tile (parameterization)
annotation = AnnotationImportanceCompass ()
tile.appendAnnotation ( annotation )
tile.draw ()
plt.show ()
print ( tile.getExplanation () )
```

## Placing Ranking Scores on the Tile

```python
parameterization = ParameterizationDefault ()
tile = Tile (parameterization)
for score in [
    RankingScore.getTrueNegativeRate (),
    RankingScore.getTruePositiveRate (),
    RankingScore.getF ( beta=0.5 ),
    RankingScore.getF ( beta=1.0 ),
    RankingScore.getF ( beta=2.0 ),
    RankingScore.getInverseF ( beta=0.5 ),
    RankingScore.getInverseF ( beta=1.0 ),
    RankingScore.getInverseF ( beta=2.0 ),
    RankingScore.getAccuracy (),
    RankingScore.getNegativePredictiveValue (),
    RankingScore.getPositivePredictiveValue ()
] :
    tile.appendAnnotation ( AnnotationText ( score ) )
tile.draw ()
plt.show ()
print ( tile.getExplanation () )
```
